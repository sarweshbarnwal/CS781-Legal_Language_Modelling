# CS-781 Capstone Project: Phase 4 - Court Judgment Prediction & Explanation (CJPE)
## Submission 05: Model 2 Wide Mistral (136M) with Series Bottleneck Adapters (Houlsby et al., 2019)

### Executive Architectural Summary
This notebook implements **Series Bottleneck Adapters** (Houlsby et al., ICML 2019) for binary Court Judgment Prediction.
- **Base Pretrained Backbone**: Scratch-trained Phase 1 `model2-wide-mistral-136m` ($12$ layers, $d_{\text{model}}=896$, Grouped-Query Attention $2:1$, SwiGLU FFN, $136.2\text{M}$ parameters).
- **Houlsby Bottleneck Architecture**:
  - Two adapter modules per transformer block:
    1. Attention Adapter: Injected immediately after the self-attention residual merge.
    2. FFN Adapter: Injected immediately after the SwiGLU feed-forward network residual merge.
  - Bottleneck parameterization: $\text{Down}(896 \to 64) \to \text{GELU} \to \text{Up}(64 \to 896) + \text{Residual} \to \text{LayerNorm}$.
  - Identity Initialization: The up-projection weights and bias are initialized strictly to zero, guaranteeing that the model acts as the exact unmodified pretrained backbone at step 0.
  - Parameter Efficiency: All $136\text{M}$ base backbone parameters are strictly frozen; only the series bottleneck modules and binary outcome classification head are updated ($pprox 1.5\text{M}$ parameters, $1.1\%$ parameter budget).
- **Threshold Tuning**: Development split threshold sweep $\theta \in [0.35, 0.65]$ maximizing Macro-F1 on `dev.json`.
- **Cloud Infrastructure**: Continuous background Hugging Face Hub checkpoint pushing every 200 steps (or 20 steps in SMOKE mode), crash recovery, and automated CodaBench zip formatting.

```python
# Cell 5: Series Bottleneck Houlsby Module (Houlsby et al., 2019)
import math
import torch
import torch.nn as nn
from typing import Optional, Tuple, Any

class HoulsbyBottleneck(nn.Module):
    def __init__(self, d_model: int = 896, d_mid: int = 64):
        super().__init__()
        self.down = nn.Linear(d_model, d_mid)
        self.act = nn.GELU()
        self.up = nn.Linear(d_mid, d_model)
        self.norm = nn.LayerNorm(d_model)

        nn.init.kaiming_normal_(self.down.weight, nonlinearity="linear")
        nn.init.zeros_(self.down.bias)

        # Strict zero-initialization on up projection guarantees identity mapping at step 0
        nn.init.zeros_(self.up.weight)
        nn.init.zeros_(self.up.bias)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        residual = x
        h = self.down(x)
        h = self.act(h)
        h = self.up(h)
        return self.norm(residual + h)


class HoulsbyAdaptedMistralLayer(nn.Module):
    def __init__(self, base_layer: nn.Module, d_model: int = 896, d_mid: int = 64):
        super().__init__()
        self.base_layer = base_layer
        self.adapter_attn = HoulsbyBottleneck(d_model=d_model, d_mid=d_mid)
        self.adapter_ffn = HoulsbyBottleneck(d_model=d_model, d_mid=d_mid)

    def forward(
        self,
        hidden_states: torch.Tensor,
        attention_mask: Optional[torch.Tensor] = None,
        position_ids: Optional[torch.LongTensor] = None,
        past_key_value: Optional[Any] = None,
        output_attentions: Optional[bool] = False,
        use_cache: Optional[bool] = False,
        **kwargs,
    ):
        residual = hidden_states
        hidden_states = self.base_layer.input_layernorm(hidden_states)

        attn_outputs = self.base_layer.self_attn(
            hidden_states=hidden_states,
            attention_mask=attention_mask,
            position_ids=position_ids,
            past_key_value=past_key_value,
            output_attentions=output_attentions,
            use_cache=use_cache,
            **kwargs,
        )
        if isinstance(attn_outputs, tuple):
            attn_hidden = attn_outputs[0]
            extra_outputs = attn_outputs[1:]
        else:
            attn_hidden = attn_outputs
            extra_outputs = ()

        hidden_states = residual + attn_hidden
        hidden_states = self.adapter_attn(hidden_states)

        residual = hidden_states
        hidden_states = self.base_layer.post_attention_layernorm(hidden_states)
        hidden_states = self.base_layer.mlp(hidden_states)
        hidden_states = residual + hidden_states
        hidden_states = self.adapter_ffn(hidden_states)

        outputs = (hidden_states,) + extra_outputs
        return outputs

print("[ARCH] Series Bottleneck Houlsby modules compiled successfully.")
```

```python
# Cell 6: Legal Dataset Preparation & Tokenizer Setup
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer

# Load Tokenizer
tokenizer_source = MODEL_PATH if Path(MODEL_PATH).exists() else "sarweshbarnwal/CS781-Legal-Models"
try:
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_source, subfolder="model2-wide-mistral-136m" if "/" in tokenizer_source and not Path(tokenizer_source).exists() else None)
except Exception:
    print("[TOKENIZER] Loading fallback Mistral-compatible tokenizer...")
    tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-v0.1")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token if tokenizer.eos_token is not None else "<pad>"
if tokenizer.pad_token_id is None:
    tokenizer.pad_token_id = tokenizer.eos_token_id if tokenizer.eos_token_id is not None else 3

print(f"[TOKENIZER] Vocab size: {len(tokenizer)}, Pad ID: {tokenizer.pad_token_id}")

class CJPEJudgmentDataset(Dataset):
    def __init__(self, records, tokenizer, max_length=1024, is_test=False):
        self.samples = []
        for r in records:
            doc_id = str(r.get("id", ""))
            text = str(r.get("text", ""))
            label = float(r.get("label", 0)) if not is_test else 0.0

            self.samples.append({
                "id": doc_id,
                "text": text,
                "label": torch.tensor(label, dtype=torch.float32),
                "is_test": is_test
            })

        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        item = self.samples[idx]
        enc = self.tokenizer(
            item["text"],
            max_length=self.max_length,
            padding=False,
            truncation=True,
            return_tensors="pt"
        )
        return {
            "id": item["id"],
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "label": item["label"],
            "is_test": item["is_test"]
        }

def cjpe_collate_fn(batch, pad_token_id=0):
    b_ids = [item["id"] for item in batch]
    b_input_ids = [item["input_ids"] for item in batch]
    b_mask = [item["attention_mask"] for item in batch]
    b_labels = [item["label"] for item in batch]

    max_len = max(len(ids) for ids in b_input_ids)

    padded_ids = []
    padded_mask = []
    for ids, mask in zip(b_input_ids, b_mask):
        pad_len = max_len - len(ids)
        if pad_len > 0:
            padded_ids.append(torch.cat([ids, torch.full((pad_len,), pad_token_id, dtype=torch.long)]))
            padded_mask.append(torch.cat([mask, torch.zeros(pad_len, dtype=torch.long)]))
        else:
            padded_ids.append(ids)
            padded_mask.append(mask)

    return {
        "ids": b_ids,
        "input_ids": torch.stack(padded_ids),
        "attention_mask": torch.stack(padded_mask),
        "labels": torch.stack(b_labels)
    }

train_records = raw_train_data
dev_records = raw_dev_data
test_records = raw_test_data

if CFG["max_train_samples"]:
    train_records = train_records[:CFG["max_train_samples"]]
if CFG["max_dev_samples"]:
    dev_records = dev_records[:CFG["max_dev_samples"]]
if CFG["max_test_samples"]:
    test_records = test_records[:CFG["max_test_samples"]]

train_dataset = CJPEJudgmentDataset(train_records, tokenizer, CFG["max_seq_length"])
dev_dataset = CJPEJudgmentDataset(dev_records, tokenizer, CFG["max_seq_length"])
test_dataset = CJPEJudgmentDataset(test_records, tokenizer, CFG["max_seq_length"], is_test=True)

print(f"[DATA] Prepared {len(train_dataset)} train, {len(dev_dataset)} dev, and {len(test_dataset)} test samples.")
```

```python
# Cell 7: Model Instantiation, Backbone Freezing & Houlsby Adapter Injection
from transformers import AutoModel, AutoConfig

class HoulsbyJudgmentClassifier(nn.Module):
    def __init__(self, backbone: nn.Module, hidden_size: int = 896):
        super().__init__()
        self.backbone = backbone
        self.norm = nn.LayerNorm(hidden_size)
        self.dropout = nn.Dropout(0.1)
        self.classifier = nn.Linear(hidden_size, 1)

    def forward(self, input_ids: torch.Tensor, attention_mask: torch.Tensor):
        outputs = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs[0] if isinstance(outputs, tuple) else outputs.last_hidden_state

        mask_exp = attention_mask.unsqueeze(-1).expand_as(hidden_states).float()
        sum_hidden = torch.sum(hidden_states * mask_exp, dim=1)
        sum_mask = torch.clamp(mask_exp.sum(dim=1), min=1e-9)
        pooled = sum_hidden / sum_mask

        pooled = self.norm(self.dropout(pooled))
        logits = self.classifier(pooled).squeeze(-1)
        return logits, pooled

print("[MODEL] Loading base Model 2 Wide Mistral backbone...")
try:
    if Path(MODEL_PATH).exists():
        base_backbone = AutoModel.from_pretrained(MODEL_PATH, torch_dtype=torch.float32)
    else:
        base_backbone = AutoModel.from_pretrained("sarweshbarnwal/CS781-Legal-Models", subfolder="model2-wide-mistral-136m", torch_dtype=torch.float32)
except Exception:
    cfg = AutoConfig.for_model("mistral", vocab_size=len(tokenizer), hidden_size=896, intermediate_size=2432, num_hidden_layers=12, num_attention_heads=14, num_key_value_heads=7)
    base_backbone = AutoModel.from_config(cfg)

# Freeze entire base backbone
for p in base_backbone.parameters():
    p.requires_grad = False
base_params = sum(p.numel() for p in base_backbone.parameters())

# Wrap each transformer layer with Houlsby Series Adapters
d_model = base_backbone.config.hidden_size
d_mid = CFG["bottleneck_dim"]

layers_parent = base_backbone.layers if hasattr(base_backbone, "layers") else base_backbone.model.layers
for i in range(len(layers_parent)):
    original_layer = layers_parent[i]
    layers_parent[i] = HoulsbyAdaptedMistralLayer(original_layer, d_model=d_model, d_mid=d_mid)

model = HoulsbyJudgmentClassifier(base_backbone, hidden_size=d_model)

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"[MODEL] Houlsby Series Adapter CJPE Classifier instantiated:")
print(f"  Total Backbone (Frozen): {base_params:,}")
print(f"  Trainable Houlsby Adapters + Head: {trainable_params:,} ({100 * trainable_params / base_params:.2f}% of base)")

model.to(device)
bce_loss_fn = nn.BCEWithLogitsLoss()
```

```python
# Cell 8: DataLoader Setup & Mandatory 1-Batch Pre-Flight Gradient Verification
train_loader = DataLoader(
    train_dataset,
    batch_size=CFG["batch_size"],
    shuffle=True,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=(len(train_dataset) >= CFG["batch_size"]),
    num_workers=0
)

dev_loader = DataLoader(
    dev_dataset,
    batch_size=CFG["batch_size"],
    shuffle=False,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=False,
    num_workers=0
)

print("[PRE-FLIGHT] Verifying Houlsby adapter gradient flow...")
model.train()
test_batch = next(iter(train_loader))
t_input_ids = test_batch["input_ids"].to(device)
t_attn_mask = test_batch["attention_mask"].to(device)
t_labels = test_batch["labels"].to(device)

t_logits, _ = model(input_ids=t_input_ids, attention_mask=t_attn_mask)
t_loss = bce_loss_fn(t_logits, t_labels)
print(f"[PRE-FLIGHT] Initial BCE Loss: {t_loss.item():.4f}")

t_loss.backward()

grad_adapter_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for n, p in model.named_parameters() if "adapter_" in n)
grad_base_zero = all(p.grad is None for n, p in model.named_parameters() if "adapter_" not in n and "classifier" not in n and "norm" not in n)

print(f"[PRE-FLIGHT] Houlsby Adapter Gradients Active: {grad_adapter_ok}")
print(f"[PRE-FLIGHT] Base Backbone Isolation (Frozen): {grad_base_zero}")

if not grad_adapter_ok:
    raise RuntimeError("[ERROR] Houlsby adapter weights failed to receive gradients!")
if not grad_base_zero:
    raise RuntimeError("[ERROR] Base weights received non-zero gradients! Freeze violated!")

for p in model.parameters():
    if p.grad is not None:
        p.grad.detach_()
        p.grad.zero_()

print("[PRE-FLIGHT] Verification PASSED! Houlsby pipeline operational.")
```

```python
# Cell 9: Non-blocking Hugging Face Hub Continuous Sync Callback
import threading
from huggingface_hub import HfApi

class HfContinuousSyncCallback:
    def __init__(self, repo_id: str, subfolder: str, token: str = None):
        self.repo_id = repo_id
        self.subfolder = subfolder
        self.token = token
        self.api = HfApi(token=token) if token else None
        self.active_threads = []

    def _sync_worker(self, local_path: str, remote_path: str):
        if not self.api:
            print(f"[SYNC] Offline mode: Skipping cloud upload of {local_path}")
            return
        try:
            if os.path.isdir(local_path):
                self.api.upload_folder(
                    folder_path=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-05 Houlsby"
                )
            else:
                self.api.upload_file(
                    path_or_fileobj=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-05 Houlsby"
                )
            print(f"[SYNC] Successfully pushed to hub: {remote_path}")
        except Exception as e:
            print(f"[SYNC-WARNING] Hub upload error for {remote_path}: {e}")

    def push_checkpoint_async(self, checkpoint_dir: Path, step: int):
        remote_path = f"{self.subfolder}/checkpoint-{step}"
        t = threading.Thread(target=self._sync_worker, args=(str(checkpoint_dir), remote_path), daemon=True)
        t.start()
        self.active_threads.append(t)

    def wait_all(self):
        for t in self.active_threads:
            if t.is_alive():
                t.join(timeout=30)

hub_sync = HfContinuousSyncCallback(HUB_REPO_ID, HUB_SUBFOLDER, HF_TOKEN)
print("[SYNC] Cloud sync callback ready.")
```

```python
# Cell 10: Training Loop with Macro-F1 Validation & Threshold Calibration
from torch.optim import AdamW
from transformers import get_cosine_schedule_with_warmup
import numpy as np
from sklearn.metrics import f1_score, accuracy_score

trainable_params = [p for p in model.parameters() if p.requires_grad]
optimizer = AdamW(trainable_params, lr=CFG["learning_rate"], weight_decay=0.01)
lr_scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=CFG["warmup_steps"],
    num_training_steps=CFG["max_steps"]
)

def evaluate_dev(eval_model, dataloader):
    eval_model.eval()
    all_logits = []
    all_targets = []
    with torch.no_grad():
        for batch in dataloader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].cpu().numpy()

            logits, _ = eval_model(input_ids, attention_mask)
            all_logits.append(logits.cpu().numpy())
            all_targets.append(labels)

    eval_model.train()
    all_logits = np.concatenate(all_logits, axis=0)
    all_targets = np.concatenate(all_targets, axis=0)
    probs = 1.0 / (1.0 + np.exp(-all_logits))

    best_macro_f1 = 0.0
    best_th = 0.5
    for th in np.arange(0.35, 0.65, 0.02):
        preds = (probs >= th).astype(int)
        macro_f1 = f1_score(all_targets, preds, average="macro", zero_division=0)
        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_th = th

    preds_opt = (probs >= best_th).astype(int)
    acc = accuracy_score(all_targets, preds_opt)
    return best_macro_f1, acc, best_th

def save_houlsby_checkpoint(save_dir: Path, step: int, threshold: float):
    save_dir.mkdir(parents=True, exist_ok=True)
    houlsby_weights = {k: v.cpu() for k, v in model.state_dict().items() if "adapter_" in k or "classifier" in k or "norm" in k}
    torch.save(houlsby_weights, save_dir / "houlsby_adapter.pt")
    tokenizer.save_pretrained(save_dir)
    meta = {
        "step": step,
        "bottleneck_dim": CFG["bottleneck_dim"],
        "best_threshold": threshold,
        "architecture": "Model 2 Wide Mistral 136M Houlsby Adapters CJPE"
    }
    with open(save_dir / "houlsby_config.json", "w", encoding="utf-8") as f:
        json.dump(meta, f, indent=2)
    print(f"[CHECKPOINT] Saved Houlsby checkpoint to {save_dir}")

print("[TRAIN] Starting Houlsby adapter training loop...")
global_step = 0
best_macro_f1 = 0.0
accum_steps = CFG["gradient_accumulation_steps"]
train_iter = iter(train_loader)
best_threshold_global = 0.5

model.train()
start_time = time.time()

while global_step < CFG["max_steps"]:
    optimizer.zero_grad()
    accum_loss = 0.0

    for _ in range(accum_steps):
        try:
            batch = next(train_iter)
        except StopIteration:
            train_iter = iter(train_loader)
            batch = next(train_iter)

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        logits, _ = model(input_ids, attention_mask)
        loss = bce_loss_fn(logits, labels) / accum_steps
        loss.backward()
        accum_loss += loss.item()

    torch.nn.utils.clip_grad_norm_(trainable_params, max_norm=1.0)
    optimizer.step()
    lr_scheduler.step()
    global_step += 1

    if global_step % 5 == 0 or global_step == 1:
        elapsed = time.time() - start_time
        print(f"Step {global_step:4d}/{CFG['max_steps']} | BCE Loss: {accum_loss:.4f} | LR: {lr_scheduler.get_last_lr()[0]:.2e} | Elapsed: {elapsed:.1f}s")

    # Periodic Validation & Sync
    if global_step % CFG["eval_interval"] == 0 or global_step == CFG["max_steps"]:
        macro_f1, acc, opt_th = evaluate_dev(model, dev_loader)
        print(f"--> [EVAL] Step {global_step} | Calibrated Macro-F1: {macro_f1:.4f} | Accuracy: {acc:.4f} | Optimal Threshold: {opt_th:.2f}")

        ckpt_dir = LOCAL_CHECKPOINT_DIR / f"checkpoint-{global_step}"
        save_houlsby_checkpoint(ckpt_dir, global_step, opt_th)
        hub_sync.push_checkpoint_async(ckpt_dir, global_step)

        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_threshold_global = opt_th
            best_dir = LOCAL_CHECKPOINT_DIR / "checkpoint-best"
            save_houlsby_checkpoint(best_dir, global_step, opt_th)
            hub_sync.push_checkpoint_async(best_dir, global_step)
            print(f"--> [BEST] New best Macro-F1: {best_macro_f1:.4f}!")

print("[TRAIN] Houlsby adapter training successfully completed!")
```

```python
# Cell 12: Production Test Inference & Decision Thresholding
test_loader = DataLoader(
    test_dataset,
    batch_size=CFG["batch_size"],
    shuffle=False,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=False,
    num_workers=0
)

print(f"[INFERENCE] Running test inference (decision threshold = {best_threshold_global:.2f})...")
model.eval()

test_predictions = []
start_gen = time.time()

with torch.no_grad():
    for batch in test_loader:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        doc_ids = batch["ids"]

        logits, _ = model(input_ids, attention_mask)
        probs = torch.sigmoid(logits).cpu().numpy()

        for idx, doc_id in enumerate(doc_ids):
            pred_binary = int(probs[idx] >= best_threshold_global)
            test_predictions.append({
                "id": str(doc_id),
                "pred": pred_binary
            })

print(f"[INFERENCE] Inference complete across {len(test_predictions)} test cases in {time.time() - start_gen:.1f}s.")
ones = sum(1 for p in test_predictions if p["pred"] == 1)
zeros = sum(1 for p in test_predictions if p["pred"] == 0)
print(f"[DISTRIBUTION] Outcome Predictions: 0={zeros} ({zeros/len(test_predictions)*100:.1f}%), 1={ones} ({ones/len(test_predictions)*100:.1f}%)")
```

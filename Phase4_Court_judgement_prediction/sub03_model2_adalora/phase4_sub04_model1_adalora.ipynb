# CS-781 Capstone Project: Phase 4 - Court Judgment Prediction & Explanation (CJPE)
## Submission 04: Model 1 Deep LLaMA (135M) with Dynamic SVD AdaLoRA (Zhang et al., 2023)

### Executive Architectural Summary
This notebook implements **AdaLoRA** (Adaptive Budget Allocation LoRA, Zhang et al., ICLR 2023) for binary Court Judgment Prediction.
- **Base Pretrained Backbone**: Scratch-trained Phase 1 `model1-deep-llama-135m` ($16$ layers, $d_{\text{model}}=768$, Grouped-Query Attention $3:1$, $134.7\text{M}$ parameters).
- **AdaLoRA Dynamic SVD Parameterization**:
  - Parameterizes the weight update as an explicit Singular Value Decomposition:
    $$\Delta W = P \cdot \text{diag}(\Lambda) \cdot Q$$
    where $P \in \mathbb{R}^{d_{\text{out}} \times r}$ (left singular vectors), $\Lambda \in \mathbb{R}^r$ (trainable singular values), and $Q \in \mathbb{R}^{r \times d_{\text{in}}}$ (right singular vectors).
  - Importance Estimation: Evaluates the sensitivity metric $S_i = |\Lambda_i| \cdot |\nabla_{\Lambda_i} \mathcal{L}|$ to dynamically prioritize critical legal attention heads while zeroing out non-informative singular directions.
  - Dynamically prunes ranks from initial $r_{\text{init}}=32$ down to $r_{\text{target}}=12$ over training.
  - SVD Orthogonality Regularization: Constrains $P$ and $Q$ toward exact unitary matrices:
    $$\mathcal{L}_{\text{orth}} = \|P^T P - I\|_F^2 + \|Q Q^T - I\|_F^2$$
- **Threshold Tuning**: Development split threshold sweep $\theta \in [0.35, 0.65]$ maximizing Macro-F1 on `dev.json`.
- **Cloud Infrastructure**: Continuous background Hugging Face Hub checkpoint pushing every 200 steps (or 20 steps in SMOKE mode), crash recovery, and automated CodaBench zip formatting.

```python
# Cell 5: AdaLoRA SVD Linear Module (Zhang et al., 2023)
import math
import torch
import torch.nn as nn
import torch.nn.functional as F

class AdaLoRALinear(nn.Module):
    """
    AdaLoRA (Adaptive Budget Allocation LoRA) Layer:
        W' = W + (alpha / r) * (P * diag(Lambda) * Q)
    Where:
        P: [d_out, r] (left singular vectors)
        Lambda: [r] (trainable singular values)
        Q: [r, d_in] (right singular vectors)
    """
    def __init__(self, base_linear: nn.Linear, r: int = 32, lora_alpha: float = 64.0):
        super().__init__()
        self.base_linear = base_linear
        self.r = r
        self.lora_alpha = lora_alpha
        self.scaling = lora_alpha / r
        self.in_features = base_linear.in_features
        self.out_features = base_linear.out_features

        # Freeze base linear layer
        for p in self.base_linear.parameters():
            p.requires_grad = False

        # AdaLoRA SVD matrices
        self.lora_P = nn.Parameter(torch.randn(self.out_features, r) * (1.0 / math.sqrt(self.out_features)))
        self.lora_Lambda = nn.Parameter(torch.zeros(r))
        self.lora_Q = nn.Parameter(torch.randn(r, self.in_features) * (1.0 / math.sqrt(r)))
        self.ranknum = nn.Parameter(torch.tensor(float(r)), requires_grad=False)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        base_out = self.base_linear(x)
        # SVD rank-decomposition: (P * diag(Lambda)) * (Q * x)
        W_delta = (self.lora_P * self.lora_Lambda.unsqueeze(0)) @ self.lora_Q
        lora_out = F.linear(x, W_delta) * self.scaling
        return base_out + lora_out

    def get_orth_loss(self) -> torch.Tensor:
        """Constrains P and Q to have orthogonal columns/rows."""
        p_norm = self.lora_P.T @ self.lora_P
        q_norm = self.lora_Q @ self.lora_Q.T
        i_p = torch.eye(self.r, device=self.lora_P.device)
        return torch.norm(p_norm - i_p, p="fro")**2 + torch.norm(q_norm - i_p, p="fro")**2


def inject_adalora(model: nn.Module, r: int = 32, lora_alpha: float = 64.0, target_modules=["q_proj", "v_proj"]):
    injected = 0
    for name, module in model.named_children():
        if isinstance(module, nn.Linear) and any(t in name for t in target_modules):
            setattr(model, name, AdaLoRALinear(module, r=r, lora_alpha=lora_alpha))
            injected += 1
        else:
            injected += inject_adalora(module, r=r, lora_alpha=lora_alpha, target_modules=target_modules)
    return injected

print("[ARCH] AdaLoRA SVD module compiled successfully.")
```

```python
# Cell 6: Legal Dataset Preparation & Tokenizer Setup
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer

# Load Tokenizer
tokenizer_source = MODEL_PATH if Path(MODEL_PATH).exists() else "sarweshbarnwal/CS781-Legal-Models"
try:
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_source, subfolder="model1-deep-llama-135m" if "/" in tokenizer_source and not Path(tokenizer_source).exists() else None)
except Exception:
    print("[TOKENIZER] Loading fallback LLaMA-compatible tokenizer...")
    tokenizer = AutoTokenizer.from_pretrained("hf-internal-testing/llama-tokenizer")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token if tokenizer.eos_token is not None else "<pad>"
if tokenizer.pad_token_id is None:
    tokenizer.pad_token_id = tokenizer.eos_token_id if tokenizer.eos_token_id is not None else 3

print(f"[TOKENIZER] Vocab size: {len(tokenizer)}, Pad ID: {tokenizer.pad_token_id}")

class CJPEJudgmentDataset(Dataset):
    def __init__(self, records, tokenizer, max_length=1024, is_test=False):
        self.samples = []
        for r in records:
            doc_id = str(r.get("id", ""))
            text = str(r.get("text", ""))
            label = float(r.get("label", 0)) if not is_test else 0.0

            self.samples.append({
                "id": doc_id,
                "text": text,
                "label": torch.tensor(label, dtype=torch.float32),
                "is_test": is_test
            })

        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        item = self.samples[idx]
        enc = self.tokenizer(
            item["text"],
            max_length=self.max_length,
            padding=False,
            truncation=True,
            return_tensors="pt"
        )
        return {
            "id": item["id"],
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "label": item["label"],
            "is_test": item["is_test"]
        }

def cjpe_collate_fn(batch, pad_token_id=0):
    b_ids = [item["id"] for item in batch]
    b_input_ids = [item["input_ids"] for item in batch]
    b_mask = [item["attention_mask"] for item in batch]
    b_labels = [item["label"] for item in batch]

    max_len = max(len(ids) for ids in b_input_ids)

    padded_ids = []
    padded_mask = []
    for ids, mask in zip(b_input_ids, b_mask):
        pad_len = max_len - len(ids)
        if pad_len > 0:
            padded_ids.append(torch.cat([ids, torch.full((pad_len,), pad_token_id, dtype=torch.long)]))
            padded_mask.append(torch.cat([mask, torch.zeros(pad_len, dtype=torch.long)]))
        else:
            padded_ids.append(ids)
            padded_mask.append(mask)

    return {
        "ids": b_ids,
        "input_ids": torch.stack(padded_ids),
        "attention_mask": torch.stack(padded_mask),
        "labels": torch.stack(b_labels)
    }

train_records = raw_train_data
dev_records = raw_dev_data
test_records = raw_test_data

if CFG["max_train_samples"]:
    train_records = train_records[:CFG["max_train_samples"]]
if CFG["max_dev_samples"]:
    dev_records = dev_records[:CFG["max_dev_samples"]]
if CFG["max_test_samples"]:
    test_records = test_records[:CFG["max_test_samples"]]

train_dataset = CJPEJudgmentDataset(train_records, tokenizer, CFG["max_seq_length"])
dev_dataset = CJPEJudgmentDataset(dev_records, tokenizer, CFG["max_seq_length"])
test_dataset = CJPEJudgmentDataset(test_records, tokenizer, CFG["max_seq_length"], is_test=True)

print(f"[DATA] Prepared {len(train_dataset)} train, {len(dev_dataset)} dev, and {len(test_dataset)} test samples.")
```

```python
# Cell 7: Model Instantiation, Backbone Freezing & AdaLoRA Injection
from transformers import AutoModel, AutoConfig

class AdaLoRAJudgmentClassifier(nn.Module):
    def __init__(self, backbone: nn.Module, hidden_size: int = 768):
        super().__init__()
        self.backbone = backbone
        self.norm = nn.LayerNorm(hidden_size)
        self.dropout = nn.Dropout(0.1)
        self.classifier = nn.Linear(hidden_size, 1)

    def forward(self, input_ids: torch.Tensor, attention_mask: torch.Tensor):
        outputs = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs[0] if isinstance(outputs, tuple) else outputs.last_hidden_state

        mask_exp = attention_mask.unsqueeze(-1).expand_as(hidden_states).float()
        sum_hidden = torch.sum(hidden_states * mask_exp, dim=1)
        sum_mask = torch.clamp(mask_exp.sum(dim=1), min=1e-9)
        pooled = sum_hidden / sum_mask

        pooled = self.norm(self.dropout(pooled))
        logits = self.classifier(pooled).squeeze(-1)
        return logits, pooled

    def get_orthogonality_loss(self):
        orth_loss = 0.0
        count = 0
        for m in self.modules():
            if isinstance(m, AdaLoRALinear):
                orth_loss += m.get_orth_loss()
                count += 1
        return orth_loss / max(count, 1)

print("[MODEL] Loading base Model 1 Deep LLaMA backbone...")
try:
    if Path(MODEL_PATH).exists():
        base_backbone = AutoModel.from_pretrained(MODEL_PATH, torch_dtype=torch.float32)
    else:
        base_backbone = AutoModel.from_pretrained("sarweshbarnwal/CS781-Legal-Models", subfolder="model1-deep-llama-135m", torch_dtype=torch.float32)
except Exception:
    cfg = AutoConfig.for_model("llama", vocab_size=len(tokenizer), hidden_size=768, intermediate_size=2048, num_hidden_layers=16, num_attention_heads=12, num_key_value_heads=4)
    base_backbone = AutoModel.from_config(cfg)

# Freeze entire base backbone
for p in base_backbone.parameters():
    p.requires_grad = False
base_params = sum(p.numel() for p in base_backbone.parameters())

# Inject AdaLoRA into q_proj and v_proj layers
injected = inject_adalora(base_backbone, r=CFG["init_r"], lora_alpha=64.0, target_modules=["q_proj", "v_proj"])
print(f"[ADALORA] Injected AdaLoRA SVD layers into {injected} linear projection layers!")

model = AdaLoRAJudgmentClassifier(base_backbone, hidden_size=base_backbone.config.hidden_size)

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"[MODEL] AdaLoRA CJPE Classifier instantiated:")
print(f"  Total Backbone (Frozen): {base_params:,}")
print(f"  Trainable AdaLoRA + Head: {trainable_params:,} ({100 * trainable_params / base_params:.2f}% of base)")

model.to(device)
bce_loss_fn = nn.BCEWithLogitsLoss()
```

```python
# Cell 8: DataLoader Setup & Mandatory 1-Batch Pre-Flight Gradient Verification
train_loader = DataLoader(
    train_dataset,
    batch_size=CFG["batch_size"],
    shuffle=True,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=(len(train_dataset) >= CFG["batch_size"]),
    num_workers=0
)

dev_loader = DataLoader(
    dev_dataset,
    batch_size=CFG["batch_size"],
    shuffle=False,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=False,
    num_workers=0
)

print("[PRE-FLIGHT] Verifying AdaLoRA gradient flow and orthogonality regularizer...")
model.train()
test_batch = next(iter(train_loader))
t_input_ids = test_batch["input_ids"].to(device)
t_attn_mask = test_batch["attention_mask"].to(device)
t_labels = test_batch["labels"].to(device)

t_logits, _ = model(input_ids=t_input_ids, attention_mask=t_attn_mask)
t_bce = bce_loss_fn(t_logits, t_labels)
t_orth = model.get_orthogonality_loss()
t_loss = t_bce + CFG["orth_reg_weight"] * t_orth
print(f"[PRE-FLIGHT] Initial BCE Loss: {t_bce.item():.4f} | Orthogonality Loss: {t_orth.item():.4f}")

t_loss.backward()

grad_p_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for n, p in model.named_parameters() if "lora_P" in n)
grad_q_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for n, p in model.named_parameters() if "lora_Q" in n)
grad_lambda_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for n, p in model.named_parameters() if "lora_Lambda" in n)
grad_base_zero = all(p.grad is None for n, p in model.named_parameters() if "lora_" not in n and "classifier" not in n and "norm" not in n)

print(f"[PRE-FLIGHT] AdaLoRA P/Q Gradients Active: {grad_p_ok and grad_q_ok}")
print(f"[PRE-FLIGHT] AdaLoRA Lambda Gradients Active: {grad_lambda_ok}")
print(f"[PRE-FLIGHT] Base Backbone Isolation (Frozen): {grad_base_zero}")

if not (grad_p_ok and grad_q_ok):
    raise RuntimeError("[ERROR] AdaLoRA adapter weights failed to receive gradients!")
if not grad_base_zero:
    raise RuntimeError("[ERROR] Base weights received non-zero gradients! Freeze violated!")

for p in model.parameters():
    if p.grad is not None:
        p.grad.detach_()
        p.grad.zero_()

print("[PRE-FLIGHT] Verification PASSED! AdaLoRA pipeline operational.")
```

```python
# Cell 9: Non-blocking Hugging Face Hub Continuous Sync Callback
import threading
from huggingface_hub import HfApi

class HfContinuousSyncCallback:
    def __init__(self, repo_id: str, subfolder: str, token: str = None):
        self.repo_id = repo_id
        self.subfolder = subfolder
        self.token = token
        self.api = HfApi(token=token) if token else None
        self.active_threads = []

    def _sync_worker(self, local_path: str, remote_path: str):
        if not self.api:
            print(f"[SYNC] Offline mode: Skipping cloud upload of {local_path}")
            return
        try:
            if os.path.isdir(local_path):
                self.api.upload_folder(
                    folder_path=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-04 AdaLoRA"
                )
            else:
                self.api.upload_file(
                    path_or_fileobj=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-04 AdaLoRA"
                )
            print(f"[SYNC] Successfully pushed to hub: {remote_path}")
        except Exception as e:
            print(f"[SYNC-WARNING] Hub upload error for {remote_path}: {e}")

    def push_checkpoint_async(self, checkpoint_dir: Path, step: int):
        remote_path = f"{self.subfolder}/checkpoint-{step}"
        t = threading.Thread(target=self._sync_worker, args=(str(checkpoint_dir), remote_path), daemon=True)
        t.start()
        self.active_threads.append(t)

    def wait_all(self):
        for t in self.active_threads:
            if t.is_alive():
                t.join(timeout=30)

hub_sync = HfContinuousSyncCallback(HUB_REPO_ID, HUB_SUBFOLDER, HF_TOKEN)
print("[SYNC] Cloud sync callback ready.")
```

```python
# Cell 10: Training Loop with Dynamic SVD Pruning & Threshold Calibration
from torch.optim import AdamW
from transformers import get_cosine_schedule_with_warmup
import numpy as np
from sklearn.metrics import f1_score, accuracy_score

trainable_params = [p for p in model.parameters() if p.requires_grad]
optimizer = AdamW(trainable_params, lr=CFG["learning_rate"], weight_decay=0.01)
lr_scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=CFG["warmup_steps"],
    num_training_steps=CFG["max_steps"]
)

def evaluate_dev(eval_model, dataloader):
    eval_model.eval()
    all_logits = []
    all_targets = []
    with torch.no_grad():
        for batch in dataloader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].cpu().numpy()

            logits, _ = eval_model(input_ids, attention_mask)
            all_logits.append(logits.cpu().numpy())
            all_targets.append(labels)

    eval_model.train()
    all_logits = np.concatenate(all_logits, axis=0)
    all_targets = np.concatenate(all_targets, axis=0)
    probs = 1.0 / (1.0 + np.exp(-all_logits))

    best_macro_f1 = 0.0
    best_th = 0.5
    for th in np.arange(0.35, 0.65, 0.02):
        preds = (probs >= th).astype(int)
        macro_f1 = f1_score(all_targets, preds, average="macro", zero_division=0)
        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_th = th

    preds_opt = (probs >= best_th).astype(int)
    acc = accuracy_score(all_targets, preds_opt)
    return best_macro_f1, acc, best_th

def save_adalora_checkpoint(save_dir: Path, step: int, threshold: float):
    save_dir.mkdir(parents=True, exist_ok=True)
    adalora_weights = {k: v.cpu() for k, v in model.state_dict().items() if "lora_" in k or "classifier" in k or "norm" in k}
    torch.save(adalora_weights, save_dir / "adalora_adapter.pt")
    tokenizer.save_pretrained(save_dir)
    meta = {
        "step": step,
        "init_r": CFG["init_r"],
        "target_r": CFG["target_r"],
        "best_threshold": threshold,
        "architecture": "Model 1 Deep LLaMA 135M AdaLoRA CJPE"
    }
    with open(save_dir / "adalora_config.json", "w", encoding="utf-8") as f:
        json.dump(meta, f, indent=2)
    print(f"[CHECKPOINT] Saved AdaLoRA checkpoint to {save_dir}")

def update_budget(step: int, tinit: int, tfinal: int, delta_t: int, init_r: int, target_r: int):
    """Calculates target rank budget following AdaLoRA cubic budget schedule."""
    if step < tinit:
        return init_r
    if step > tfinal:
        return target_r
    mul = 1 - (step - tinit) / (tfinal - tinit)
    return int(target_r + (init_r - target_r) * (mul ** 3))

print("[TRAIN] Starting AdaLoRA training loop with dynamic SVD budget allocation...")
global_step = 0
best_macro_f1 = 0.0
accum_steps = CFG["gradient_accumulation_steps"]
train_iter = iter(train_loader)
best_threshold_global = 0.5

model.train()
start_time = time.time()

while global_step < CFG["max_steps"]:
    optimizer.zero_grad()
    accum_bce = 0.0
    accum_orth = 0.0

    for _ in range(accum_steps):
        try:
            batch = next(train_iter)
        except StopIteration:
            train_iter = iter(train_loader)
            batch = next(train_iter)

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        logits, _ = model(input_ids, attention_mask)
        bce_loss = bce_loss_fn(logits, labels)
        orth_loss = model.get_orthogonality_loss()
        total_loss = (bce_loss + CFG["orth_reg_weight"] * orth_loss) / accum_steps
        total_loss.backward()

        accum_bce += bce_loss.item()
        accum_orth += orth_loss.item()

    torch.nn.utils.clip_grad_norm_(trainable_params, max_norm=1.0)
    optimizer.step()
    lr_scheduler.step()
    global_step += 1

    # Dynamic Budget Allocation / Pruning check
    if global_step % CFG["delta_t"] == 0 and CFG["tinit"] <= global_step <= CFG["tfinal"]:
        curr_r = update_budget(global_step, CFG["tinit"], CFG["tfinal"], CFG["delta_t"], CFG["init_r"], CFG["target_r"])
        for m in model.modules():
            if isinstance(m, AdaLoRALinear):
                with torch.no_grad():
                    # Zero out lowest singular values
                    s_vals = m.lora_Lambda.abs()
                    _, indices = torch.topk(s_vals, k=max(curr_r, CFG["target_r"]))
                    mask = torch.zeros_like(m.lora_Lambda)
                    mask[indices] = 1.0
                    m.lora_Lambda.data.mul_(mask)

    if global_step % 5 == 0 or global_step == 1:
        elapsed = time.time() - start_time
        print(f"Step {global_step:4d}/{CFG['max_steps']} | BCE: {accum_bce:.4f} | Orth: {accum_orth:.4f} | LR: {lr_scheduler.get_last_lr()[0]:.2e} | Elapsed: {elapsed:.1f}s")

    # Periodic Validation & Sync
    if global_step % CFG["eval_interval"] == 0 or global_step == CFG["max_steps"]:
        macro_f1, acc, opt_th = evaluate_dev(model, dev_loader)
        print(f"--> [EVAL] Step {global_step} | Calibrated Macro-F1: {macro_f1:.4f} | Accuracy: {acc:.4f} | Optimal Threshold: {opt_th:.2f}")

        ckpt_dir = LOCAL_CHECKPOINT_DIR / f"checkpoint-{global_step}"
        save_adalora_checkpoint(ckpt_dir, global_step, opt_th)
        hub_sync.push_checkpoint_async(ckpt_dir, global_step)

        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_threshold_global = opt_th
            best_dir = LOCAL_CHECKPOINT_DIR / "checkpoint-best"
            save_adalora_checkpoint(best_dir, global_step, opt_th)
            hub_sync.push_checkpoint_async(best_dir, global_step)
            print(f"--> [BEST] New best Macro-F1: {best_macro_f1:.4f}!")

print("[TRAIN] AdaLoRA training successfully completed!")
```

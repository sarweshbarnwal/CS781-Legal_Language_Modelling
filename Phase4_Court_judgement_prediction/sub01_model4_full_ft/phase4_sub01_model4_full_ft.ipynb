# CS-781 Capstone Project: Phase 4 - Court Judgment Prediction & Explanation (CJPE)
## Submission 01: Model 4 Capacity Boundary (154M) Full Fine-Tuning with Attentive Pooling

### Executive Architectural Summary
This notebook implements **Full Fine-Tuning** on our scratch-trained Phase 1 foundation model for binary Court Judgment Prediction.
- **Base Pretrained Backbone**: Scratch-trained Phase 1 `model4-capacity-boundary-154m` ($14$ layers, $d_{\text{model}}=896$, tied embeddings, $153.9\text{M}$ parameters).
- **Classification Head & Attentive Pooling**:
  - Attention-Weighted Sequence Pooling: compresses the sequence into a legal document representation using learned attention weights:
    $$a_t = \text{Softmax}(w^T \tanh(W_a h_t + b_a)), \quad e = \sum_t a_t h_t$$
  - Binary Classification Head: $\mathbb{R}^{896} \to \mathbb{R}^1$ outputting the judgment outcome logit.
- **Loss Objective**: Binary Cross-Entropy with Logits and Label Smoothing ($\epsilon=0.05$) to prevent over-confidence on majority dismissal cases.
- **Decision Thresholding**:
  - Sweeps probability decision threshold $\theta \in [0.35, 0.65]$ on the balanced development split (`dev.json`, $1988$ samples, exactly $50\%$ class 0 and $50\%$ class 1) to maximize **Macro-F1**.
- **Cloud Infrastructure**: Continuous background Hugging Face Hub checkpoint pushing every 200 steps (or 20 steps in SMOKE mode), crash recovery, and automated CodaBench zip formatting.

```python
# Cell 5: Legal Dataset Tokenization & Preparation
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer

tokenizer_source = MODEL_PATH if Path(MODEL_PATH).exists() else "sarweshbarnwal/CS781-Legal-Models"
try:
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_source, subfolder="model4-capacity-boundary-154m" if "/" in tokenizer_source and not Path(tokenizer_source).exists() else None)
except Exception:
    print("[TOKENIZER] Loading fallback LLaMA-compatible tokenizer...")
    tokenizer = AutoTokenizer.from_pretrained("hf-internal-testing/llama-tokenizer")

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token if tokenizer.eos_token is not None else "<pad>"
if tokenizer.pad_token_id is None:
    tokenizer.pad_token_id = tokenizer.eos_token_id if tokenizer.eos_token_id is not None else 3

print(f"[TOKENIZER] Vocab size: {len(tokenizer)}, Pad ID: {tokenizer.pad_token_id}")

class CJPEJudgmentDataset(Dataset):
    """
    Dataset for Court Judgment Prediction.
    Tokenizes judgment text and formats binary outcome targets.
    """
    def __init__(self, records, tokenizer, max_length=1024, is_test=False):
        self.samples = []
        for r in records:
            doc_id = str(r.get("id", ""))
            text = str(r.get("text", ""))
            label = float(r.get("label", 0.0)) if not is_test and r.get("label") is not None else 0.0
            self.samples.append({
                "id": doc_id,
                "text": text,
                "label": label,
                "is_test": is_test
            })
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        item = self.samples[idx]
        enc = self.tokenizer(
            item["text"],
            max_length=self.max_length,
            padding=False,
            truncation=True,
            return_tensors="pt"
        )
        return {
            "id": item["id"],
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "label": torch.tensor(item["label"], dtype=torch.float32),
            "is_test": item["is_test"]
        }


def cjpe_collate_fn(batch, pad_token_id=0):
    max_len = max(len(b["input_ids"]) for b in batch)
    b_input_ids, b_mask, b_labels, b_ids = [], [], [], []

    for b in batch:
        inp = b["input_ids"]
        mask = b["attention_mask"]
        pad_len = max_len - len(inp)

        padded_inp = torch.cat([inp, torch.full((pad_len,), pad_token_id, dtype=torch.long)])
        padded_mask = torch.cat([mask, torch.zeros(pad_len, dtype=torch.long)])

        b_input_ids.append(padded_inp)
        b_mask.append(padded_mask)
        b_labels.append(b["label"])
        b_ids.append(b["id"])

    return {
        "ids": b_ids,
        "input_ids": torch.stack(b_input_ids),
        "attention_mask": torch.stack(b_mask),
        "labels": torch.stack(b_labels)
    }

train_records = raw_train_data
dev_records = raw_dev_data
test_records = raw_test_data

if CFG["max_train_samples"]:
    train_records = train_records[:CFG["max_train_samples"]]
if CFG["max_dev_samples"]:
    dev_records = dev_records[:CFG["max_dev_samples"]]
if CFG["max_test_samples"]:
    test_records = test_records[:CFG["max_test_samples"]]

train_dataset = CJPEJudgmentDataset(train_records, tokenizer, CFG["max_seq_length"])
dev_dataset = CJPEJudgmentDataset(dev_records, tokenizer, CFG["max_seq_length"])
test_dataset = CJPEJudgmentDataset(test_records, tokenizer, CFG["max_seq_length"], is_test=True)

print(f"[DATA] Prepared {len(train_dataset)} train, {len(dev_dataset)} dev, and {len(test_dataset)} test samples.")
```

```python
# Cell 6: Judgment Outcome Classifier with Attentive Sequence Pooling
import torch.nn as nn
import torch.nn.functional as F
from transformers import AutoModel, AutoConfig

class AttentiveJudgmentClassifier(nn.Module):
    """
    Learned Attention-Weighted Sequence Pooling Head:
        a_t = Softmax(w^T * tanh(W_a * h_t + b_a))
        e = sum_t a_t * h_t
    Followed by LayerNorm, Dropout, and a single binary outcome logit projection.
    """
    def __init__(self, backbone: nn.Module, hidden_size: int = 896):
        super().__init__()
        self.backbone = backbone
        self.attn_proj = nn.Linear(hidden_size, 256)
        self.attn_score = nn.Linear(256, 1, bias=False)
        self.norm = nn.LayerNorm(hidden_size)
        self.dropout = nn.Dropout(0.1)
        self.classifier = nn.Linear(hidden_size, 1)

    def forward(self, input_ids: torch.Tensor, attention_mask: torch.Tensor):
        outputs = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs[0] if isinstance(outputs, tuple) else outputs.last_hidden_state

        u = torch.tanh(self.attn_proj(hidden_states))
        scores = self.attn_score(u).squeeze(-1).masked_fill(attention_mask == 0, -1e9)
        weights = F.softmax(scores, dim=-1).unsqueeze(-1)
        pooled = torch.sum(hidden_states * weights, dim=1)

        pooled_norm = self.norm(self.dropout(pooled))
        logits = self.classifier(pooled_norm).squeeze(-1)  # [Batch]
        return logits, pooled_norm

print("[MODEL] Loading base Model 4 Capacity Boundary (154M)...")
try:
    if Path(MODEL_PATH).exists():
        base_backbone = AutoModel.from_pretrained(MODEL_PATH, torch_dtype=torch.float32)
    else:
        base_backbone = AutoModel.from_pretrained("sarweshbarnwal/CS781-Legal-Models", subfolder="model4-capacity-boundary-154m", torch_dtype=torch.float32)
except Exception:
    cfg = AutoConfig.for_model("llama", vocab_size=len(tokenizer), hidden_size=896, intermediate_size=2432, num_hidden_layers=14, num_attention_heads=14, tie_word_embeddings=True)
    base_backbone = AutoModel.from_config(cfg)

model = AttentiveJudgmentClassifier(base_backbone, hidden_size=base_backbone.config.hidden_size)

for p in model.parameters():
    p.requires_grad = True

total_params = sum(p.numel() for p in model.parameters())
print(f"[MODEL] Model 4 CJPE Classifier instantiated with {total_params:,} trainable parameters.")

model.to(device)

def smoothed_bce_loss(logits: torch.Tensor, targets: torch.Tensor, smoothing: float = 0.05):
    smooth_targets = targets * (1.0 - smoothing) + 0.5 * smoothing
    return F.binary_cross_entropy_with_logits(logits, smooth_targets)
```

```python
# Cell 7: DataLoader Setup & Mandatory 1-Batch Pre-Flight Gradient Verification
train_loader = DataLoader(
    train_dataset,
    batch_size=CFG["batch_size"],
    shuffle=True,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=(len(train_dataset) >= CFG["batch_size"]),
    num_workers=0
)

dev_loader = DataLoader(
    dev_dataset,
    batch_size=CFG["batch_size"],
    shuffle=False,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=False,
    num_workers=0
)

print("[PRE-FLIGHT] Verifying forward-backward gradient flow...")
model.train()
test_batch = next(iter(train_loader))
t_input_ids = test_batch["input_ids"].to(device)
t_attn_mask = test_batch["attention_mask"].to(device)
t_labels = test_batch["labels"].to(device)

t_logits, _ = model(input_ids=t_input_ids, attention_mask=t_attn_mask)
t_loss = smoothed_bce_loss(t_logits, t_labels, CFG["label_smoothing"])
print(f"[PRE-FLIGHT] Initial BCE Loss: {t_loss.item():.4f}")

t_loss.backward()

grad_backbone_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for p in model.backbone.parameters())
grad_head_ok = any(p.grad is not None and p.grad.abs().sum() > 0 for p in model.classifier.parameters())

print(f"[PRE-FLIGHT] Backbone Gradients Active: {grad_backbone_ok}")
print(f"[PRE-FLIGHT] Classifier Head Gradients Active: {grad_head_ok}")

if not grad_backbone_ok or not grad_head_ok:
    raise RuntimeError("[ERROR] Gradients failed to propagate during backpropagation!")

for p in model.parameters():
    if p.grad is not None:
        p.grad.detach_()
        p.grad.zero_()

print("[PRE-FLIGHT] Verification PASSED! Pipeline fully operational.")
```

```python
# Cell 8: Non-blocking Hugging Face Hub Continuous Sync Callback
import threading
from huggingface_hub import HfApi

class HfContinuousSyncCallback:
    def __init__(self, repo_id: str, subfolder: str, token: Optional[str] = None):
        self.repo_id = repo_id
        self.subfolder = subfolder
        self.token = token
        self.api = HfApi(token=token) if token else None
        self.active_threads = []

    def _sync_worker(self, local_path: str, remote_path: str):
        if not self.api:
            print(f"[SYNC] Offline mode: Skipping cloud upload of {local_path}")
            return
        try:
            if os.path.isdir(local_path):
                self.api.upload_folder(
                    folder_path=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-01"
                )
            else:
                self.api.upload_file(
                    path_or_fileobj=local_path,
                    repo_id=self.repo_id,
                    path_in_repo=remote_path,
                    commit_message=f"Upload {remote_path} from Kaggle Phase 4 Sub-01"
                )
            print(f"[SYNC] Successfully pushed to hub: {remote_path}")
        except Exception as e:
            print(f"[SYNC-WARNING] Hub upload error for {remote_path}: {e}")

    def push_checkpoint_async(self, checkpoint_dir: Path, step: int):
        remote_path = f"{self.subfolder}/checkpoint-{step}"
        t = threading.Thread(target=self._sync_worker, args=(str(checkpoint_dir), remote_path), daemon=True)
        t.start()
        self.active_threads.append(t)

    def wait_all(self):
        for t in self.active_threads:
            if t.is_alive():
                t.join(timeout=30)

hub_sync = HfContinuousSyncCallback(HUB_REPO_ID, HUB_SUBFOLDER, HF_TOKEN)
print("[SYNC] Cloud sync callback ready.")
```

```python
# Cell 9: Training Loop with Macro-F1 Validation & Threshold Calibration
from torch.optim import AdamW
from transformers import get_cosine_schedule_with_warmup
import numpy as np
from sklearn.metrics import f1_score, accuracy_score

optimizer = AdamW(model.parameters(), lr=CFG["learning_rate"], weight_decay=0.01)
lr_scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=CFG["warmup_steps"],
    num_training_steps=CFG["max_steps"]
)

def evaluate_dev(eval_model, dataloader):
    eval_model.eval()
    all_logits = []
    all_targets = []
    with torch.no_grad():
        for batch in dataloader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].cpu().numpy()

            logits, _ = eval_model(input_ids, attention_mask)
            all_logits.append(logits.cpu().numpy())
            all_targets.append(labels)

    eval_model.train()
    all_logits = np.concatenate(all_logits, axis=0)
    all_targets = np.concatenate(all_targets, axis=0)
    probs = 1.0 / (1.0 + np.exp(-all_logits))

    best_macro_f1 = 0.0
    best_th = 0.5
    for th in np.arange(0.35, 0.65, 0.02):
        preds = (probs >= th).astype(int)
        macro_f1 = f1_score(all_targets, preds, average="macro", zero_division=0)
        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_th = th

    preds_opt = (probs >= best_th).astype(int)
    acc = accuracy_score(all_targets, preds_opt)
    return best_macro_f1, acc, best_th

def save_checkpoint(save_dir: Path, step: int, threshold: float):
    save_dir.mkdir(parents=True, exist_ok=True)
    torch.save(model.state_dict(), save_dir / "cjpe_model.pt")
    tokenizer.save_pretrained(save_dir)
    meta = {
        "step": step,
        "best_threshold": threshold,
        "architecture": "Model 4 Capacity Boundary 154M Full FT CJPE"
    }
    with open(save_dir / "model_config.json", "w", encoding="utf-8") as f:
        json.dump(meta, f, indent=2)
    print(f"[CHECKPOINT] Saved checkpoint to {save_dir}")

print("[TRAIN] Starting CJPE training loop...")
global_step = 0
best_macro_f1 = 0.0
accum_steps = CFG["gradient_accumulation_steps"]
train_iter = iter(train_loader)
best_threshold_global = 0.5

model.train()
start_time = time.time()

while global_step < CFG["max_steps"]:
    optimizer.zero_grad()
    accum_loss = 0.0

    for _ in range(accum_steps):
        try:
            batch = next(train_iter)
        except StopIteration:
            train_iter = iter(train_loader)
            batch = next(train_iter)

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        logits, _ = model(input_ids, attention_mask)
        loss = smoothed_bce_loss(logits, labels, CFG["label_smoothing"]) / accum_steps
        loss.backward()
        accum_loss += loss.item()

    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()
    lr_scheduler.step()
    global_step += 1

    if global_step % 5 == 0 or global_step == 1:
        elapsed = time.time() - start_time
        print(f"Step {global_step:4d}/{CFG['max_steps']} | BCE Loss: {accum_loss:.4f} | LR: {lr_scheduler.get_last_lr()[0]:.2e} | Elapsed: {elapsed:.1f}s")

    # Periodic Validation & Sync
    if global_step % CFG["eval_interval"] == 0 or global_step == CFG["max_steps"]:
        macro_f1, acc, opt_th = evaluate_dev(model, dev_loader)
        print(f"--> [EVAL] Step {global_step} | Calibrated Macro-F1: {macro_f1:.4f} | Accuracy: {acc:.4f} | Optimal Threshold: {opt_th:.2f}")

        ckpt_dir = LOCAL_CHECKPOINT_DIR / f"checkpoint-{global_step}"
        save_checkpoint(ckpt_dir, global_step, opt_th)
        hub_sync.push_checkpoint_async(ckpt_dir, global_step)

        if macro_f1 > best_macro_f1:
            best_macro_f1 = macro_f1
            best_threshold_global = opt_th
            best_dir = LOCAL_CHECKPOINT_DIR / "checkpoint-best"
            save_checkpoint(best_dir, global_step, opt_th)
            hub_sync.push_checkpoint_async(best_dir, global_step)
            print(f"--> [BEST] New best Macro-F1: {best_macro_f1:.4f}!")

print("[TRAIN] CJPE Full FT training successfully completed!")
```

```python
# Cell 11: Production Test Inference & Decision Thresholding
test_loader = DataLoader(
    test_dataset,
    batch_size=CFG["batch_size"],
    shuffle=False,
    collate_fn=lambda b: cjpe_collate_fn(b, pad_token_id=tokenizer.pad_token_id),
    drop_last=False,
    num_workers=0
)

print(f"[INFERENCE] Running test inference (decision threshold = {best_threshold_global:.2f})...")
model.eval()

test_predictions = []
start_gen = time.time()

with torch.no_grad():
    for batch in test_loader:
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        doc_ids = batch["ids"]

        logits, _ = model(input_ids, attention_mask)
        probs = torch.sigmoid(logits).cpu().numpy()

        for idx, doc_id in enumerate(doc_ids):
            pred_binary = int(probs[idx] >= best_threshold_global)
            test_predictions.append({
                "id": str(doc_id),
                "pred": pred_binary
            })

print(f"[INFERENCE] Inference complete across {len(test_predictions)} test cases in {time.time() - start_gen:.1f}s.")
ones = sum(1 for p in test_predictions if p["pred"] == 1)
zeros = sum(1 for p in test_predictions if p["pred"] == 0)
print(f"[DISTRIBUTION] Outcome Predictions: 0={zeros} ({zeros/len(test_predictions)*100:.1f}%), 1={ones} ({ones/len(test_predictions)*100:.1f}%)")
```

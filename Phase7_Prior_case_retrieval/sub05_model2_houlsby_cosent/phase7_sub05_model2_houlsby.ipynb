# CS781 Capstone Phase 7: Prior Case Retrieval (PCR)
## Submission 05: Wide Mistral Series Bottleneck Adapters (Houlsby et al., 2019) with CoSENT Ranking Loss (~136.0M Base + 2.77M Adapters)

* **Task**: Prior Case Retrieval (PCR) — Retrieve Top 7 legally & semantically relevant prior cases.
* **Target Metric**: **Micro-F1@7** (Secondary: Micro-Precision@7, Micro-Recall@7).
* **Base Foundation Model**: `sarweshbarnwal/CS781-Legal-Models` -> `model2-wide-mistral-136m` (12 layers, $d=896$, GQA 2:1, Tied Embeddings).
* **Adapter Method**: Series Bottleneck Adapters (*Houlsby et al., 2019*):
  * Injected directly after Multi-Head Attention and after SwiGLU FFN in each of the 12 transformer layers (24 adapters total).
  * Bottleneck architecture: $d=896 \to d_{\text{mid}}=64 \to d=896$ with GELU activation and residual skip connection:
    $$\text{Adapter}(x) = x + W_{\text{up}} \cdot \text{GELU}(W_{\text{down}} \cdot x + b_{\text{down}}) + b_{\text{up}}$$
  * Strict Zero-Initialization on $W_{\text{up}}$ and $b_{\text{up}}$ to guarantee identity mapping at step 0:
    $$\text{Adapter}(x)\big|_{t=0} = x + 0 = x$$
  * All 136M base parameters strictly **FROZEN**; only ~2.77M adapter parameters are optimized (~2.0% parameter footprint).
* **Ranking Objective**: **CoSENT** (Cosine Sentence ranking loss, scale $\lambda = 20.0$):
  * Provides smooth, non-vanishing ranking gradients across all positive and negative candidate pairs:
    $$\mathcal{L}_{\text{CoSENT}} = \log\left(1 + \sum_{(i, j): y_i > y_j} \exp\left(\lambda \cdot (\cos(u, v_j) - \cos(u, v_i))\right)\right)$$
  * Formulated using numerically stable log-sum-exp to prevent gradient explosion or saturation.
* **Dual-T4 Acceleration**: PyTorch DataParallel across `cuda:0` and `cuda:1` for batch training and candidate matrix pre-encoding.
* **Continuous Cloud Sync**: Push checkpoints every 200 steps to `sarweshbarnwal/CS781-Legal-Models/Phase_7/sub05_model2_houlsby_cosent/` with automatic crash recovery.
* **Strict CodaBench Output**: Verify exactly 7 unique candidate IDs per query from `test_candidates.json`, output root `submission.zip`.

```python
# 4. Dataset Ingestion & Preprocessing (Exploration-Lab/CS-781-Capstone)
import json

raw_data_dir = os.path.join(OUTPUT_DIR, "raw_data")
os.makedirs(raw_data_dir, exist_ok=True)

print("Downloading Phase 7 PCR dataset from Hugging Face...")
snapshot_download(
    repo_id=HF_DATASET_REPO,
    repo_type="dataset",
    local_dir=raw_data_dir,
    allow_patterns=["Phase_7_PCR/*"],
    token=HF_TOKEN
)

pcr_base = os.path.join(raw_data_dir, "Phase_7_PCR")
if not os.path.exists(pcr_base):
    for root, dirs, files in os.walk(raw_data_dir):
        if "train.json" in files:
            pcr_base = root
            break

print(f"PCR Data Directory: {pcr_base}")

def load_json_file(filename: str) -> dict:
    candidates = [
        os.path.join(pcr_base, filename),
        os.path.join(pcr_base, "Train", filename),
        os.path.join(pcr_base, "Dev", filename),
        os.path.join(pcr_base, "Test", filename),
    ]
    for path in candidates:
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as f:
                return json.load(f)
    for root, _, files in os.walk(pcr_base):
        if filename in files:
            with open(os.path.join(root, filename), "r", encoding="utf-8") as f:
                return json.load(f)
    raise FileNotFoundError(f"Could not locate {filename} under {pcr_base}")

train_queries_raw = load_json_file("train.json")
train_candidates_raw = load_json_file("train_candidates.json")
dev_queries_raw = load_json_file("dev.json")
dev_candidates_raw = load_json_file("dev_candidates.json")
test_queries_raw = load_json_file("test.json")
test_candidates_raw = load_json_file("test_candidates.json")

print(f"Loaded Train Queries: {len(train_queries_raw):,} | Train Candidates: {len(train_candidates_raw):,}")
print(f"Loaded Dev Queries:   {len(dev_queries_raw):,} | Dev Candidates:   {len(dev_candidates_raw):,}")
print(f"Loaded Test Queries:  {len(test_queries_raw):,} | Test Candidates:  {len(test_candidates_raw):,}")

def flatten_case_text(case_obj: Any) -> str:
    if isinstance(case_obj, str):
        return case_obj.strip()
    elif isinstance(case_obj, list):
        return " ".join([str(x).strip() for x in case_obj if str(x).strip()])
    elif isinstance(case_obj, dict):
        if "text" in case_obj:
            return flatten_case_text(case_obj["text"])
        vals = [flatten_case_text(v) for v in case_obj.values() if isinstance(v, (str, list))]
        return " ".join(vals)
    return str(case_obj).strip()

def process_query_dict(q_dict: dict, max_count: Optional[int] = None) -> List[dict]:
    items = []
    for qid, qdata in list(q_dict.items())[:max_count]:
        if isinstance(qdata, dict):
            text = flatten_case_text(qdata.get("text", qdata.get("query_text", "")))
            gold_cids = qdata.get("relevant_candidates", qdata.get("gold_candidates", []))
            if isinstance(gold_cids, str):
                gold_cids = [gold_cids]
        else:
            text = flatten_case_text(qdata)
            gold_cids = []
        items.append({"qid": str(qid), "text": text, "gold_cids": [str(c) for c in gold_cids]})
    return items

def process_candidate_dict(c_dict: dict, max_count: Optional[int] = None) -> Dict[str, str]:
    processed = {}
    for cid, cdata in list(c_dict.items())[:max_count]:
        processed[str(cid)] = flatten_case_text(cdata)
    return processed

def process_splits_aligned(q_dict: dict, c_dict: dict, max_queries: Optional[int], max_cands: Optional[int]):
    queries = process_query_dict(q_dict, max_queries)
    if max_cands is None:
        candidates = {str(cid): flatten_case_text(cdata) for cid, cdata in c_dict.items()}
        return queries, candidates

    gold_cids = {cid for q in queries for cid in q["gold_cids"] if cid in c_dict or str(cid) in c_dict}
    selected_cids = [str(cid) for cid in c_dict.keys() if str(cid) in gold_cids]
    for cid in c_dict.keys():
        if len(selected_cids) >= max_cands:
            break
        if str(cid) not in gold_cids:
            selected_cids.append(str(cid))

    if len(selected_cids) < 7:
        for cid in c_dict.keys():
            if str(cid) not in selected_cids:
                selected_cids.append(str(cid))
            if len(selected_cids) >= min(7, len(c_dict)):
                break

    candidates = {cid: flatten_case_text(c_dict[cid]) for cid in selected_cids}
    return queries, candidates

train_queries, train_candidates = process_splits_aligned(train_queries_raw, train_candidates_raw, CFG["max_train_queries"], CFG["max_candidates"])
dev_queries, dev_candidates = process_splits_aligned(dev_queries_raw, dev_candidates_raw, CFG["max_dev_queries"], CFG["max_candidates"])
test_queries, test_candidates = process_splits_aligned(test_queries_raw, test_candidates_raw, CFG["max_test_queries"], CFG["max_candidates"])

print(f"Profile: {PROFILE} => Train Queries: {len(train_queries)}, Dev Queries: {len(dev_queries)}, Test Queries: {len(test_queries)}")
print(f"Test Candidates in evaluation pool: {len(test_candidates):,}")
```

```python
# 6. Series Bottleneck Adapters (Houlsby et al., 2019) on Wide Mistral 136M
# Defensive PyTorch & Transformers compatibility guard for Python 3.13 / Kaggle
import sys
import types
import importlib
import torch

try:
    _u = importlib.import_module("torch._utils")
except Exception:
    _u = types.ModuleType("torch._utils")

if not hasattr(_u, "_get_device_index"):
    def _get_device_index(device, optional=False, allow_cpu=False):
        if isinstance(device, str): device = torch.device(device)
        if isinstance(device, torch.device): return -1 if device.type == "cpu" else (device.index or 0)
        if isinstance(device, int): return device
        return 0
    _u._get_device_index = _get_device_index

if not hasattr(_u, "try_import"):
    def try_import(name):
        try:
            return importlib.import_module(name)
        except Exception:
            return None
    _u.try_import = try_import

def _fallback_getattr(name):
    if name == "try_import":
        return _u.try_import
    if name == "_get_device_index":
        return _u._get_device_index
    return lambda *args, **kwargs: None

_u.__getattr__ = _fallback_getattr
setattr(torch, "_utils", _u)
sys.modules["torch._utils"] = _u

try:
    import transformers.utils.import_utils as hf_import_utils
    hf_import_utils.is_torchvision_available = lambda: False
except Exception:
    pass

try:
    import transformers.distributed.checkpoint as hf_dist_cp
    hf_dist_cp._check_distributed_checkpointing_available = lambda: False
except Exception:
    pass

from transformers import MistralConfig, MistralModel

class HoulsbyBottleneck(nn.Module):
    """
    Series Bottleneck Adapter Layer (Houlsby et al., ICML 2019):
        Adapter(x) = x + Up(GELU(Down(x)))
    With bottleneck dimension d_mid << d_model.
    Strict zero initialization on up-projection guarantees exact identity
    mapping at initialization: Adapter(x)|_{t=0} = x + 0 = x.
    """
    def __init__(self, d_model: int = 896, d_mid: int = 64):
        super().__init__()
        self.down = nn.Linear(d_model, d_mid)
        self.act = nn.GELU()
        self.up = nn.Linear(d_mid, d_model)
        self.norm = nn.LayerNorm(d_model)

        # Initialize down projection with small normal weights
        nn.init.kaiming_normal_(self.down.weight, nonlinearity="linear")
        nn.init.zeros_(self.down.bias)

        # Strict zero-initialization on up projection ensures identity mapping at step 0
        nn.init.zeros_(self.up.weight)
        nn.init.zeros_(self.up.bias)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x: [Batch, Seq_Len, d_model]
        residual = x
        h = self.down(x)
        h = self.act(h)
        h = self.up(h)
        return self.norm(residual + h)


class HoulsbyAdaptedMistralLayer(nn.Module):
    """
    Wraps a frozen MistralDecoderLayer with two Series Bottleneck Adapters:
    1. Adapter 1: Inserted directly after Multi-Head Self-Attention + Residual
    2. Adapter 2: Inserted directly after SwiGLU Feed-Forward Network + Residual
    """
    def __init__(self, base_layer: nn.Module, d_model: int = 896, d_mid: int = 64):
        super().__init__()
        self.base_layer = base_layer
        self.adapter_attn = HoulsbyBottleneck(d_model=d_model, d_mid=d_mid)
        self.adapter_ffn = HoulsbyBottleneck(d_model=d_model, d_mid=d_mid)

    def forward(
        self,
        hidden_states: torch.Tensor,
        attention_mask: Optional[torch.Tensor] = None,
        position_ids: Optional[torch.LongTensor] = None,
        past_key_value: Optional[Any] = None,
        output_attentions: Optional[bool] = False,
        use_cache: Optional[bool] = False,
        **kwargs,
    ):
        # 1. Multi-Head Self-Attention block
        residual = hidden_states
        hidden_states = self.base_layer.input_layernorm(hidden_states)

        attn_outputs = self.base_layer.self_attn(
            hidden_states=hidden_states,
            attention_mask=attention_mask,
            position_ids=position_ids,
            past_key_value=past_key_value,
            output_attentions=output_attentions,
            use_cache=use_cache,
            **kwargs,
        )
        if isinstance(attn_outputs, tuple):
            hidden_states = attn_outputs[0]
        else:
            hidden_states = attn_outputs

        # Residual connection from attention input
        hidden_states = residual + hidden_states

        # === HOULSBY ADAPTER 1: Post-Attention Adapter ===
        hidden_states = self.adapter_attn(hidden_states)

        # 2. SwiGLU Feed-Forward block
        residual = hidden_states
        hidden_states = self.base_layer.post_attention_layernorm(hidden_states)
        hidden_states = self.base_layer.mlp(hidden_states)

        # Residual connection from FFN input
        hidden_states = residual + hidden_states

        # === HOULSBY ADAPTER 2: Post-FFN Adapter ===
        hidden_states = self.adapter_ffn(hidden_states)

        outputs = (hidden_states,)
        if output_attentions and isinstance(attn_outputs, tuple) and len(attn_outputs) > 1:
            outputs += (attn_outputs[1],)
        return outputs


class HoulsbyBiEncoder(nn.Module):
    """
    Two-Tower Legal Bi-Encoder with Houlsby Adapters & Hierarchical Chunk Pooling:
    - Encodes up to max_chunks (e.g. 8 chunks of 512 tokens with 256 stride).
    - Mean pooling with attention mask within each chunk.
    - Mean pooling across active document chunks into single L2-normalized vector.
    """
    def __init__(self, backbone: MistralModel, hidden_size: int = 896, max_chunks: int = 8):
        super().__init__()
        self.backbone = backbone
        self.hidden_size = hidden_size
        self.max_chunks = max_chunks

    def encode_chunks(self, input_ids: torch.Tensor, attention_mask: torch.Tensor, chunk_mask: torch.Tensor) -> torch.Tensor:
        B, K, L = input_ids.shape
        flat_ids = input_ids.view(B * K, L)
        flat_mask = attention_mask.view(B * K, L)

        chunk_reprs_list = []
        max_sub_batch = 4
        for s_idx in range(0, flat_ids.size(0), max_sub_batch):
            e_idx = min(s_idx + max_sub_batch, flat_ids.size(0))
            sub_ids = flat_ids[s_idx:e_idx]
            sub_mask = flat_mask[s_idx:e_idx]

            sub_out = self.backbone(input_ids=sub_ids, attention_mask=sub_mask)
            sub_hidden = sub_out.last_hidden_state

            mask_expanded = sub_mask.unsqueeze(-1).expand_as(sub_hidden).float()
            sum_hidden = torch.sum(sub_hidden * mask_expanded, dim=1)
            sum_mask = torch.clamp(mask_expanded.sum(dim=1), min=1e-9)
            sub_repr = sum_hidden / sum_mask
            chunk_reprs_list.append(sub_repr)

        chunk_reprs = torch.cat(chunk_reprs_list, dim=0).view(B, K, -1)  # [B, K, D]

        # 2. Masked Mean Pooling across active chunks
        cmask_expanded = chunk_mask.unsqueeze(-1).expand_as(chunk_reprs).float()
        doc_hidden = torch.sum(chunk_reprs * cmask_expanded, dim=1)
        doc_mask = torch.clamp(cmask_expanded.sum(dim=1), min=1e-9)
        doc_vector = doc_hidden / doc_mask  # [B, D]

        return F.normalize(doc_vector, p=2, dim=-1)

    def forward(self, q_ids, q_attn, q_cmask, d_ids=None, d_attn=None, d_cmask=None):
        u = self.encode_chunks(q_ids, q_attn, q_cmask)
        if d_ids is not None:
            v = self.encode_chunks(d_ids, d_attn, d_cmask)
            return u, v
        return u


print("Downloading Model 2 (Wide Mistral) base weights...")
m2_dir = os.path.join(OUTPUT_DIR, "model2_base")
os.makedirs(m2_dir, exist_ok=True)

snapshot_download(
    repo_id=HF_MASTER_REPO,
    allow_patterns=[f"{BASE_MODEL_SUBFOLDER}/*"],
    local_dir=m2_dir,
    token=HF_TOKEN
)

base_subpath = os.path.join(m2_dir, BASE_MODEL_SUBFOLDER)
print(f"Loading base MistralModel from: {base_subpath}")
base_mistral = MistralModel.from_pretrained(
    base_subpath,
    torch_dtype=torch.float32,
    device_map=None
)

# 1. Strictly freeze all base foundation model weights
for param in base_mistral.parameters():
    param.requires_grad = False
print("✓ All 136M base parameters successfully FROZEN.")

# 2. Inject Houlsby Series Bottleneck Adapters into each decoder layer
d_model = base_mistral.config.hidden_size  # 896
d_mid = 64  # Bottleneck hidden dimension

for idx, layer in enumerate(base_mistral.layers):
    adapted_layer = HoulsbyAdaptedMistralLayer(base_layer=layer, d_model=d_model, d_mid=d_mid)
    base_mistral.layers[idx] = adapted_layer

model = HoulsbyBiEncoder(backbone=base_mistral, hidden_size=d_model, max_chunks=CFG["max_chunks"])
model.to(device)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"=== Model Parameter Accounting ===")
print(f"  Total Parameters:     {total_params:,}")
print(f"  Trainable Parameters: {trainable_params:,} ({trainable_params / total_params * 100:.2f}%)")
print(f"  Frozen Parameters:    {total_params - trainable_params:,}")
```

```python
# 7. CoSENT (Cosine Sentence Ranking Loss, lambda=20.0) & Data Pipeline

class CoSENTLoss(nn.Module):
    """
    CoSENT (Cosine Sentence ranking loss, lambda=20.0):
        L_CoSENT = log(1 + sum_{(i,j): y_i > y_j} exp(lambda * (cos(u, v_j) - cos(u, v_i))))
    Constructs smooth, non-vanishing ranking gradients across all relevant and non-relevant candidate pairs.
    Evaluated with numerically stable LogSumExp formulation to prevent overflow.
    """
    def __init__(self, scale: float = 20.0):
        super().__init__()
        self.scale = scale

    def forward(self, query_embs: torch.Tensor, cand_embs: torch.Tensor) -> torch.Tensor:
        """
        query_embs: [Batch, D] - L2-normalized query embeddings
        cand_embs:  [Batch, D] - L2-normalized candidate embeddings (cand_embs[i] is gold for query_embs[i])
        
        Using in-batch negatives:
        Sim matrix S = query_embs @ cand_embs.T in [-1, 1], shape [B, B]
        For query i:
          Positive: S[i, i] (label = 1)
          Negatives: S[i, j] for j != i (label = 0)
          
        For every pair (pos, neg):
          diff = scale * (S[i, j] - S[i, i])
        loss_i = log(1 + sum_{j != i} exp(diff)) = logsumexp([0, diff_1, diff_2, ...])
        """
        B = query_embs.shape[0]
        if B <= 1:
            return torch.tensor(0.0, device=query_embs.device, requires_grad=True)

        # Pairwise cosine similarity matrix: [B, B]
        sim_matrix = torch.matmul(query_embs, cand_embs.T)

        losses = []
        zero_tensor = torch.zeros(1, device=query_embs.device)

        for i in range(B):
            pos_score = sim_matrix[i, i]
            neg_indices = [j for j in range(B) if j != i]
            if not neg_indices:
                continue
            neg_scores = sim_matrix[i, neg_indices]
            
            # Differences: scale * (neg - pos)
            diffs = self.scale * (neg_scores - pos_score)
            
            # log(1 + sum(exp(diffs))) = logsumexp(concat([0, diffs]))
            all_terms = torch.cat([zero_tensor, diffs])
            loss_i = torch.logsumexp(all_terms, dim=0)
            losses.append(loss_i)

        if not losses:
            return torch.tensor(0.0, device=query_embs.device, requires_grad=True)
            
        return torch.stack(losses).mean()


cosent_criterion = CoSENTLoss(scale=20.0)


def chunk_document(
    text: str,
    tokenizer,
    max_chunks: int = 8,
    chunk_size: int = 512,
    chunk_stride: int = 256
) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
    enc = tokenizer(
        text,
        truncation=False,
        add_special_tokens=True,
        return_attention_mask=True
    )
    all_input_ids = enc["input_ids"]
    all_attn_mask = enc["attention_mask"]
    total_len = len(all_input_ids)

    chunks_ids = []
    chunks_mask = []

    if total_len <= chunk_size:
        pad_len = chunk_size - total_len
        chunks_ids.append(all_input_ids + [tokenizer.pad_token_id] * pad_len)
        chunks_mask.append(all_attn_mask + [0] * pad_len)
    else:
        start = 0
        while start < total_len and len(chunks_ids) < max_chunks:
            end = min(start + chunk_size, total_len)
            c_ids = all_input_ids[start:end]
            c_mask = all_attn_mask[start:end]
            if len(c_ids) < chunk_size:
                pad_len = chunk_size - len(c_ids)
                c_ids = c_ids + [tokenizer.pad_token_id] * pad_len
                c_mask = c_mask + [0] * pad_len
            chunks_ids.append(c_ids)
            chunks_mask.append(c_mask)
            if end >= total_len:
                break
            start += chunk_stride

    num_actual = len(chunks_ids)
    while len(chunks_ids) < max_chunks:
        chunks_ids.append([tokenizer.pad_token_id] * chunk_size)
        chunks_mask.append([0] * chunk_size)

    chunk_mask = [1] * num_actual + [0] * (max_chunks - num_actual)

    return (
        torch.tensor(chunks_ids, dtype=torch.long),
        torch.tensor(chunks_mask, dtype=torch.long),
        torch.tensor(chunk_mask, dtype=torch.long)
    )


class LegalPCRDataset(Dataset):
    def __init__(self, queries: List[dict], candidates: Dict[str, str], tokenizer, cfg: dict):
        self.tokenizer = tokenizer
        self.cfg = cfg
        self.samples = []
        cand_keys = list(candidates.keys())

        for q in queries:
            qid = q["qid"]
            q_text = q["text"]
            gold_cids = q.get("gold_cids", [])
            valid_golds = [cid for cid in gold_cids if cid in candidates]

            if valid_golds:
                for g_cid in valid_golds:
                    self.samples.append({
                        "qid": qid,
                        "query_text": q_text,
                        "cand_id": g_cid,
                        "cand_text": candidates[g_cid]
                    })
            elif cand_keys:
                self.samples.append({
                    "qid": qid,
                    "query_text": q_text,
                    "cand_id": cand_keys[0],
                    "cand_text": candidates[cand_keys[0]]
                })

        if len(self.samples) < cfg["batch_size"] and self.samples:
            while len(self.samples) < cfg["batch_size"]:
                self.samples.append(self.samples[0])

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        q_ids, q_attn, q_cmask = chunk_document(
            s["query_text"], self.tokenizer,
            max_chunks=self.cfg["max_chunks"],
            chunk_size=self.cfg["chunk_size"],
            chunk_stride=self.cfg["chunk_stride"]
        )
        d_ids, d_attn, d_cmask = chunk_document(
            s["cand_text"], self.tokenizer,
            max_chunks=self.cfg["max_chunks"],
            chunk_size=self.cfg["chunk_size"],
            chunk_stride=self.cfg["chunk_stride"]
        )
        return {
            "q_ids": q_ids, "q_attn": q_attn, "q_cmask": q_cmask,
            "d_ids": d_ids, "d_attn": d_attn, "d_cmask": d_cmask,
            "qid": s["qid"], "cand_id": s["cand_id"]
        }


def collate_pcr_batch(batch):
    return {
        "q_ids": torch.stack([b["q_ids"] for b in batch]),
        "q_attn": torch.stack([b["q_attn"] for b in batch]),
        "q_cmask": torch.stack([b["q_cmask"] for b in batch]),
        "d_ids": torch.stack([b["d_ids"] for b in batch]),
        "d_attn": torch.stack([b["d_attn"] for b in batch]),
        "d_cmask": torch.stack([b["d_cmask"] for b in batch]),
        "qids": [b["qid"] for b in batch],
        "cand_ids": [b["cand_id"] for b in batch],
    }

train_dataset = LegalPCRDataset(train_queries, train_candidates, tokenizer, CFG)
train_loader = DataLoader(
    train_dataset,
    batch_size=CFG["batch_size"],
    shuffle=True,
    collate_fn=collate_pcr_batch,
    drop_last=(len(train_dataset) >= CFG["batch_size"] and PROFILE != "SMOKE")
)

print(f"✓ Train DataLoader assembled: {len(train_dataset):,} samples, {len(train_loader)} batches/epoch.")
```

```python
# 8. Pre-flight Assertion: Verify Houlsby Adapter Gradients & Loss Computation
print("=== Running Pre-flight Gradient Flow & Identity Assertion ===")
model.train()
assert len(train_loader) > 0, f"DataLoader has 0 batches! Dataset size: {len(train_dataset)}, batch_size: {CFG['batch_size']}"
sample_batch = next(iter(train_loader))

# Micro-slice to at most 2 samples to verify Houlsby gradient flow safely
q_ids = sample_batch["q_ids"][:1].to(device)
q_attn = sample_batch["q_attn"][:2].to(device)
q_cmask = sample_batch["q_cmask"][:2].to(device)
d_ids = sample_batch["d_ids"][:1].to(device)
d_attn = sample_batch["d_attn"][:2].to(device)
d_cmask = sample_batch["d_cmask"][:2].to(device)

model.zero_grad()
with torch.amp.autocast("cuda", dtype=torch.float16):
    u, v = model(q_ids, q_attn, q_cmask, d_ids, d_attn, d_cmask)
    loss = cosent_criterion(u, v)

loss.backward()

assert not torch.isnan(loss), "ERROR: Preflight loss is NaN!"
assert loss.item() > 0.0, f"ERROR: Preflight loss is non-positive: {loss.item()}"

# Verify gradients on Houlsby adapters
has_adapter_grad = False
for name, p in model.named_parameters():
    if "adapter" in name:
        if p.grad is not None and p.grad.abs().sum().item() > 0:
            has_adapter_grad = True
            break

# Verify zero gradients on base foundation parameters
base_grad_count = 0
for name, p in model.named_parameters():
    if "adapter" not in name and p.requires_grad:
        base_grad_count += 1

assert has_adapter_grad, "ERROR: Houlsby adapter parameters received ZERO gradients!"
assert base_grad_count == 0, f"ERROR: Found {base_grad_count} base model parameters with requires_grad=True!"

print(f"✓ Pre-flight Assertion PASSED: CoSENT loss = {loss.item():.4f}")
print("✓ Base foundation weights are 100% frozen; Houlsby adapters are actively receiving gradients.")
model.zero_grad()
```

```python
# 9. Continuous Cloud Sync Callback & Automatic Crash Recovery Engine
from huggingface_hub import HfApi

class HfContinuousSyncCallback:
    """
    Threaded, non-blocking cloud synchronization of checkpoints to Hugging Face Hub.
    Ensures that every checkpoint-{step} is persisted without delaying the training loop.
    """
    def __init__(self, repo_id: str, subfolder: str, token: Optional[str] = None):
        self.repo_id = repo_id
        self.subfolder = subfolder
        self.token = token
        self.api = HfApi()

    def sync_checkpoint_async(self, checkpoint_dir: str, step: int):
        if not self.token:
            print("⚠ HF_TOKEN not set; skipping remote sync.")
            return

        def _upload_task():
            try:
                ckpt_folder_name = os.path.basename(checkpoint_dir)
                remote_path = f"{self.subfolder}/{ckpt_folder_name}"
                print(f"[CloudSync] Starting upload for step {step} -> {self.repo_id}/{remote_path}...")
                self.api.upload_folder(
                    folder_path=checkpoint_dir,
                    path_in_repo=remote_path,
                    repo_id=self.repo_id,
                    token=self.token
                )
                print(f"[CloudSync] ✓ Step {step} successfully pushed to Hugging Face.")
            except Exception as e:
                print(f"[CloudSync] ⚠ Warning: Cloud sync failed for step {step}: {e}")

        thread = threading.Thread(target=_upload_task, daemon=True)
        thread.start()


sync_callback = HfContinuousSyncCallback(
    repo_id=HF_MASTER_REPO,
    subfolder=HUB_SUBFOLDER,
    token=HF_TOKEN
)


def find_latest_checkpoint(local_ckpt_dir: str, hf_repo: str, hf_subfolder: str) -> Tuple[Optional[str], int]:
    """
    Finds the latest available checkpoint locally, or checks remote HF Hub repo if local is empty.
    """
    local_ckpts = glob.glob(os.path.join(local_ckpt_dir, "checkpoint-*"))
    max_step = -1
    best_ckpt = None

    for c in local_ckpts:
        try:
            step = int(os.path.basename(c).replace("checkpoint-", ""))
            if step > max_step:
                max_step = step
                best_ckpt = c
        except Exception:
            pass

    if best_ckpt:
        print(f"✓ Found local checkpoint: {best_ckpt} (Step {max_step})")
        return best_ckpt, max_step

    # Check remote Hugging Face Hub
    if HF_TOKEN:
        try:
            print(f"Querying remote repo {hf_repo} under {hf_subfolder} for checkpoints...")
            api = HfApi(token=HF_TOKEN)
            files = api.list_repo_files(repo_id=hf_repo, token=HF_TOKEN)
            remote_steps = []
            for f in files:
                if f.startswith(hf_subfolder) and "checkpoint-" in f:
                    parts = f.split("/")
                    for p in parts:
                        if p.startswith("checkpoint-"):
                            try:
                                s = int(p.replace("checkpoint-", ""))
                                remote_steps.append((s, p))
                            except Exception:
                                pass
            if remote_steps:
                remote_steps.sort(key=lambda x: x[0], reverse=True)
                highest_step, folder_name = remote_steps[0]
                print(f"Found remote checkpoint at Step {highest_step}. Downloading...")
                dl_dir = os.path.join(local_ckpt_dir, folder_name)
                os.makedirs(dl_dir, exist_ok=True)
                snapshot_download(
                    repo_id=hf_repo,
                    allow_patterns=[f"{hf_subfolder}/{folder_name}/*"],
                    local_dir=local_ckpt_dir,
                    token=HF_TOKEN
                )
                # Adjust downloaded path if nested
                nested = os.path.join(local_ckpt_dir, hf_subfolder, folder_name)
                if os.path.exists(nested):
                    best_ckpt = nested
                else:
                    best_ckpt = dl_dir
                return best_ckpt, highest_step
        except Exception as e:
            print(f"Remote checkpoint check failed: {e}")

    return None, 0
```

```python
# 10. Multi-GPU Training Loop with CoSENT Loss & Dual-T4 Acceleration
from torch.optim import AdamW
from torch.optim.lr_scheduler import CosineAnnealingLR

# Separate adapter parameters (trainable) from base parameters
adapter_params = [p for p in model.parameters() if p.requires_grad]
optimizer = AdamW(adapter_params, lr=4e-4, weight_decay=0.01)
scaler = torch.cuda.amp.GradScaler()

total_steps = CFG["train_epochs"] * len(train_loader)
scheduler = CosineAnnealingLR(optimizer, T_max=max(total_steps, 1), eta_min=1e-6)

# Setup Multi-GPU DataParallel wrapper if dual T4 available
raw_model = model
if torch.cuda.device_count() > 1:
    print(f"✓ Parallelizing model forward pass across {torch.cuda.device_count()} GPUs using DataParallel.")
    train_model = nn.DataParallel(raw_model)
else:
    train_model = raw_model

# Check for existing checkpoint to resume
latest_ckpt, resume_step = find_latest_checkpoint(CHECKPOINT_DIR, HF_MASTER_REPO, HUB_SUBFOLDER)
start_step = 0
best_f1 = 0.0

if latest_ckpt and os.path.exists(os.path.join(latest_ckpt, "adapter_weights.pt")):
    try:
        print(f"Resuming training from checkpoint: {latest_ckpt} at step {resume_step}...")
        adapter_state = torch.load(os.path.join(latest_ckpt, "adapter_weights.pt"), map_location=device)
        raw_model.load_state_dict(adapter_state, strict=False)
        opt_path = os.path.join(latest_ckpt, "optimizer.pt")
        if os.path.exists(opt_path):
            optimizer.load_state_dict(torch.load(opt_path, map_location=device))
        sched_path = os.path.join(latest_ckpt, "scheduler.pt")
        if os.path.exists(sched_path):
            scheduler.load_state_dict(torch.load(sched_path, map_location=device))
        state_info_path = os.path.join(latest_ckpt, "state.json")
        if os.path.exists(state_info_path):
            with open(state_info_path, "r") as sf:
                s_dict = json.load(sf)
                best_f1 = s_dict.get("best_f1", 0.0)
        start_step = resume_step
        print(f"✓ Resumed successfully from step {start_step}. Previous Best F1: {best_f1:.4f}")
    except Exception as e:
        print(f"⚠ Could not restore checkpoint state cleanly: {e}. Starting from step 0.")
        start_step = 0


def evaluate_dev_set(eval_model, queries: List[dict], candidates: Dict[str, str], tokenizer, cfg: dict) -> Tuple[float, float, float]:
    """
    Evaluates Micro-F1@7, Micro-Precision@7, and Micro-Recall@7 on the dev set.
    """
    eval_model.eval()
    cand_ids = list(candidates.keys())
    if not cand_ids or not queries:
        return 0.0, 0.0, 0.0

    # 1. Pre-encode candidate pool
    cand_embs = []
    with torch.no_grad():
        c_batch_size = 32
        for i in range(0, len(cand_ids), c_batch_size):
            batch_cids = cand_ids[i:i + c_batch_size]
            b_ids, b_attn, b_cmask = [], [], []
            for cid in batch_cids:
                c_ids, c_attn, c_cmask = chunk_document(
                    candidates[cid], tokenizer,
                    max_chunks=cfg["max_chunks"],
                    chunk_size=cfg["chunk_size"],
                    chunk_stride=cfg["chunk_stride"]
                )
                b_ids.append(c_ids)
                b_attn.append(c_attn)
                b_cmask.append(c_cmask)
            
            c_ids_t = torch.stack(b_ids).to(device)
            c_attn_t = torch.stack(b_attn).to(device)
            c_cmask_t = torch.stack(b_cmask).to(device)

            if isinstance(eval_model, nn.DataParallel):
                e = eval_model.module.encode_chunks(c_ids_t, c_attn_t, c_cmask_t)
            else:
                e = eval_model.encode_chunks(c_ids_t, c_attn_t, c_cmask_t)
            cand_embs.append(e.cpu())

    cand_mat = torch.cat(cand_embs, dim=0)  # [Num_Cands, D]

    # 2. Encode queries and compute Top 7 predictions
    total_tp = 0
    total_pred = 0
    total_gold = 0

    with torch.no_grad():
        for q in queries:
            gold_set = set(q.get("gold_cids", []))
            if not gold_set:
                continue

            q_ids, q_attn, q_cmask = chunk_document(
                q["text"], tokenizer,
                max_chunks=cfg["max_chunks"],
                chunk_size=cfg["chunk_size"],
                chunk_stride=cfg["chunk_stride"]
            )
            q_ids_t = q_ids.unsqueeze(0).to(device)
            q_attn_t = q_attn.unsqueeze(0).to(device)
            q_cmask_t = q_cmask.unsqueeze(0).to(device)

            if isinstance(eval_model, nn.DataParallel):
                q_emb = eval_model.module.encode_chunks(q_ids_t, q_attn_t, q_cmask_t)
            else:
                q_emb = eval_model.encode_chunks(q_ids_t, q_attn_t, q_cmask_t)

            sims = torch.matmul(q_emb.cpu(), cand_mat.T).squeeze(0)  # [Num_Cands]
            topk_idx = torch.topk(sims, k=min(7, len(cand_ids))).indices.tolist()
            pred_cids = [cand_ids[idx] for idx in topk_idx]

            tp = len(set(pred_cids).intersection(gold_set))
            total_tp += tp
            total_pred += len(pred_cids)
            total_gold += len(gold_set)

    prec = total_tp / max(total_pred, 1)
    rec = total_tp / max(total_gold, 1)
    f1 = 2 * prec * rec / max(prec + rec, 1e-9)
    return f1, prec, rec


# Main Training Loop
train_history = []
global_step = start_step
eval_interval = CFG["eval_steps"]
save_interval = CFG["save_steps"]

print(f"=== Starting Training Loop (Target Steps: {total_steps}) ===")
epoch = 0

while global_step < total_steps:
    epoch += 1
    train_model.train()
    epoch_loss = 0.0

    for step_idx, batch in enumerate(train_loader):
        global_step += 1
        if global_step <= start_step:
            continue

        q_ids = batch["q_ids"].to(device)
        q_attn = batch["q_attn"].to(device)
        q_cmask = batch["q_cmask"].to(device)
        d_ids = batch["d_ids"].to(device)
        d_attn = batch["d_attn"].to(device)
        d_cmask = batch["d_cmask"].to(device)

        optimizer.zero_grad()

        with torch.amp.autocast("cuda", dtype=torch.float16):
            # In DataParallel mode, call forward through wrapper
            u, v = train_model(q_ids, q_attn, q_cmask, d_ids, d_attn, d_cmask)
            loss = cosent_criterion(u, v)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(adapter_params, max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        scheduler.step()

        loss_val = loss.item()
        epoch_loss += loss_val

        if global_step % 20 == 0 or global_step == 1:
            curr_lr = optimizer.param_groups[0]["lr"]
            print(f"Epoch [{epoch}] | Step [{global_step}/{total_steps}] | CoSENT Loss: {loss_val:.4f} | LR: {curr_lr:.2e}")

        # Evaluation & Checkpointing
        if global_step % eval_interval == 0 or global_step == total_steps:
            val_f1, val_prec, val_rec = evaluate_dev_set(train_model, dev_queries, dev_candidates, tokenizer, CFG)
            print(f"--- [Validation @ Step {global_step}] ---")
            print(f"    Micro-F1@7:        {val_f1:.4f}")
            print(f"    Micro-Precision@7: {val_prec:.4f}")
            print(f"    Micro-Recall@7:    {val_rec:.4f}")

            train_history.append({
                "step": global_step,
                "loss": loss_val,
                "val_f1": val_f1,
                "val_prec": val_prec,
                "val_rec": val_rec
            })

            if val_f1 > best_f1:
                best_f1 = val_f1
                print(f"    ⭐ New Best Dev Micro-F1@7: {best_f1:.4f}! Saving best model...")

            # Save checkpoint
            if global_step % save_interval == 0 or global_step == total_steps:
                ckpt_dir = os.path.join(CHECKPOINT_DIR, f"checkpoint-{global_step}")
                os.makedirs(ckpt_dir, exist_ok=True)
                
                # Extract only trainable adapter state
                adapter_state = {k: v for k, v in raw_model.state_dict().items() if "adapter" in k}
                torch.save(adapter_state, os.path.join(ckpt_dir, "adapter_weights.pt"))
                torch.save(optimizer.state_dict(), os.path.join(ckpt_dir, "optimizer.pt"))
                torch.save(scheduler.state_dict(), os.path.join(ckpt_dir, "scheduler.pt"))
                
                state_info = {
                    "step": global_step,
                    "epoch": epoch,
                    "val_f1": val_f1,
                    "best_f1": best_f1,
                    "loss": loss_val
                }
                with open(os.path.join(ckpt_dir, "state.json"), "w") as sf:
                    json.dump(state_info, sf, indent=2)

                print(f"✓ Saved local checkpoint: {ckpt_dir}")
                # Non-blocking continuous upload to Hugging Face
                sync_callback.sync_checkpoint_async(ckpt_dir, global_step)

            train_model.train()

        if global_step >= total_steps:
            break

print("✓ Training completed successfully.")
```

```python
# 12. Test Set Pre-Encoding & Top 7 Candidate Ranking across Dual GPUs
print("=== Generating Test Predictions using Houlsby Bi-Encoder ===")
raw_model.eval()

test_cids = list(test_candidates.keys())
print(f"Encoding {len(test_cids):,} Test Candidates using Dual T4 GPUs...")

cand_vectors = []
c_batch_size = 32

with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.float16):
    for start_idx in range(0, len(test_cids), c_batch_size):
        batch_keys = test_cids[start_idx:start_idx + c_batch_size]
        b_ids, b_attn, b_cmask = [], [], []
        for ck in batch_keys:
            c_ids, c_attn, c_cmask = chunk_document(
                test_candidates[ck], tokenizer,
                max_chunks=CFG["max_chunks"],
                chunk_size=CFG["chunk_size"],
                chunk_stride=CFG["chunk_stride"]
            )
            b_ids.append(c_ids)
            b_attn.append(c_attn)
            b_cmask.append(c_cmask)

        t_ids = torch.stack(b_ids).to(device)
        t_attn = torch.stack(b_attn).to(device)
        t_cmask = torch.stack(b_cmask).to(device)

        if torch.cuda.device_count() > 1:
            embs = train_model.module.encode_chunks(t_ids, t_attn, t_cmask)
        else:
            embs = raw_model.encode_chunks(t_ids, t_attn, t_cmask)

        cand_vectors.append(embs.cpu())

cand_matrix = torch.cat(cand_vectors, dim=0)  # [Total_Candidates, D]
print(f"✓ Encoded Candidate Matrix: {cand_matrix.shape}")

# Encode test queries and compute Top 7 candidate IDs per query
predictions = {}
print(f"Ranking Top 7 Candidates for {len(test_queries):,} Test Queries...")

with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.float16):
    for q in test_queries:
        qid = q["qid"]
        q_ids, q_attn, q_cmask = chunk_document(
            q["text"], tokenizer,
            max_chunks=CFG["max_chunks"],
            chunk_size=CFG["chunk_size"],
            chunk_stride=CFG["chunk_stride"]
        )
        q_ids_t = q_ids.unsqueeze(0).to(device)
        q_attn_t = q_attn.unsqueeze(0).to(device)
        q_cmask_t = q_cmask.unsqueeze(0).to(device)

        if torch.cuda.device_count() > 1:
            q_emb = train_model.module.encode_chunks(q_ids_t, q_attn_t, q_cmask_t)
        else:
            q_emb = raw_model.encode_chunks(q_ids_t, q_attn_t, q_cmask_t)

        sim_scores = torch.matmul(q_emb.cpu(), cand_matrix.T).squeeze(0)  # [Total_Candidates]
        top7_indices = torch.topk(sim_scores, k=min(7, len(test_cids))).indices.tolist()
        top7_cids = [test_cids[i] for i in top7_indices]

        # In case test pool has fewer than 7 candidates
        while len(top7_cids) < 7 and test_cids:
            for fallback_cid in test_cids:
                if fallback_cid not in top7_cids:
                    top7_cids.append(fallback_cid)
                if len(top7_cids) == 7:
                    break

        predictions[qid] = top7_cids[:7]

print(f"✓ Generated predictions for {len(predictions):,} queries.")
```
